# Superstore Sales Analysis

A reproducible walkthrough using Python, Pandas, Matplotlib, Seaborn, and the project analysis pipeline. Run this notebook after installing the packages in `requirements.txt`. If generated data is missing, the setup cell downloads the public sample CSV and creates the cleaned outputs automatically.

```python
from pathlib import Path
import subprocess
import sys

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

def find_repo_root():
    for candidate in [Path.cwd(), *Path.cwd().parents]:
        if (candidate / 'src' / 'analyze.py').exists():
            return candidate
    raise FileNotFoundError('Run this notebook from the cloned repository so src/analyze.py can be found.')

REPO_ROOT = find_repo_root()
OUTPUT_DIR = REPO_ROOT / 'outputs'
CLEANED_DATA = OUTPUT_DIR / 'superstore_cleaned.csv'

if not CLEANED_DATA.exists():
    subprocess.run([
        sys.executable, str(REPO_ROOT / 'src' / 'analyze.py'),
        '--download', '--output', str(OUTPUT_DIR)
    ], check=True)

df = pd.read_csv(CLEANED_DATA, parse_dates=['order_date', 'ship_date'])
df.shape
```

```python
df[['order_id', 'order_date', 'sales', 'profit', 'category', 'region']].head()
```

```python
summary = (
    df.groupby('category')[['sales', 'profit']]
      .sum()
      .sort_values('profit', ascending=False)
)
summary
```

```python
plot_data = summary.reset_index()
plt.figure(figsize=(8, 4))
sns.barplot(data=plot_data, x='profit', y='category', hue='category', palette='viridis', legend=False)
plt.title('Profit by Category')
plt.xlabel('Profit')
plt.ylabel('')
plt.tight_layout()
plt.show()
```
